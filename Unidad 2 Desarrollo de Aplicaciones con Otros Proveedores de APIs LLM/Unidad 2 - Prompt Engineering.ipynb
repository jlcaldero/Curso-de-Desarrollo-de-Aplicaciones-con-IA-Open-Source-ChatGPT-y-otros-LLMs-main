## Importar librerias

```python
import os
from openai import OpenAI
from dotenv import load_dotenv, find_dotenv
```

## Configuración del Cliente de OpenAI con Variables de Entorno

```python
# Cargar variables del archivo .env
load_dotenv()

# Obtener la API key desde las variables de entorno
openai_api_key = os.getenv("OPENAI_API_KEY")

# Inicializar cliente de OpenAI
client = OpenAI(api_key=openai_api_key)
```

```python
def ChatCompletion(messages):
    response = client.chat.completions.create(
        model="gpt-4o",
        messages=messages,
        temperature=0.7,
        max_tokens=2500,
        top_p=1,
        frequency_penalty=0,
        presence_penalty=0
    )
    return response.choices[0].message.content
```

## Patrón de Persona (Prompting de Rol)

```python
messages = [
    {
        "role": "system",
        "content": "Imagina que eres Carl Sagan. Explica la teoría de la relatividad de una manera que cualquier persona pueda entender."
    }
]

ChatCompletion(messages)
```

```python
messages = [
    {
        "role": "system",
        "content": "Imagina que eres un famoso Chef, eres Donato de Santis. Proporciona una receta detallada para preparar unos penne rigate."
    }
]
ChatCompletion(messages)
```

```python
messages = [
    {
        "role": "system",
        "content": "Eres un asistente de viajes experto. Ayudas a los usuarios a planificar sus vacaciones, proporcionando recomendaciones de destinos, itinerarios y consejos de viaje."
    },
    {
        "role": "user",
        "content": "Hola, estoy planeando unas vacaciones en Europa para dos semanas. ¿Qué destinos me recomiendas visitar?"
    }
]

ChatCompletion(messages)
```

## Patrón de Refinamiento de Preguntas

```python
messages = [
    {
        "role": "user",
        "content": "Quiero desarrollar una aplicación móvil como Uber. De ahora en más quiero que me hagas 5 preguntas para entender mejor mi petición y luego de que te las responda si puedes continuar. responde ok."
    }
]

response = ChatCompletion(messages)

messages.append({
        "role": "assistant",
        "content": response
    }
)

messages
```

```python
respuestas = []
for i in range(5):
    respuesta = input(f"Respuesta a la pregunta {i+1}: ")
    respuestas.append(respuesta)
```

```python
inputs = [{"role": "user", "content": f"Respuesta a la pregunta {i+1}: {respuestas[i]}"} for i in range(5)]

messages.extend(inputs)

messages
ChatCompletion(messages)
```

## Prompting de Pocos Disparos (Few-Shot Prompting)

```python
messages = [
    {
        "role": "system",
        "content": """Ejemplo 1:
                    Petición: Dame una breve descripción de París.
                    Respuesta: París es la capital de Francia, conocida por su arte, cultura y monumentos icónicos como la Torre Eiffel y la Catedral de Notre Dame.

                    Ejemplo 2:
                    Petición: ¿Cuál es la capital de Japón?
                    Respuesta: La capital de Japón es Tokio.
                
                    Ejemplo 3:
                    Petición: ¿Puedes explicar la teoría de la relatividad?
                    Respuesta: La teoría de la relatividad, propuesta por Albert Einstein, incluye la relatividad especial y la relatividad general. La relatividad especial establece que las leyes de la física son las mismas para todos los observadores no acelerados, y que la velocidad de la luz en el vacío es constante. La relatividad general expande estos principios para incluir la gravedad como una curvatura del espacio-tiempo.
        """
    },
    {
        "role": "user",
        "content": "¿Qué es la fotosíntesis?"
    }    
]

ChatCompletion(messages)
```

## Cadena de Pensamiento (COT) de Cero Disparos (Zero Shot Chain of Thought)

```python
messages = [
    {
        "role": "system",
        "content": "Responde a la siguiente pregunta proporcionando tu razonamiento paso a paso"
    },
    {
        "role": "user",
        "content": "Si Juan tiene 3 manzanas y compra 2 más, y luego le da 1 a su amigo, ¿cuántas manzanas tiene ahora?"
    }
]

ChatCompletion(messages)
```
