## Importar librerias

```python
from langchain_openai import OpenAI
from langchain_core.prompts import PromptTemplate
from dotenv import load_dotenv
import os
```

## Configuración Variables de Entorno

```python
# Cargar variables del archivo .env
load_dotenv()

# Obtener la API key desde las variables de entorno
openai_api_key = os.getenv("OPENAI_API_KEY")
```

## Uso de PromptTemplate de LangChain para interactuar con el modelo

```python
template = """Pregunta: {question}

Respuesta: Pensemos paso a paso."""


prompt = PromptTemplate.from_template(template)

# Crear la instancia del modelo con la API key
llm = OpenAI(openai_api_key=openai_api_key)

llm_chain = prompt | llm

# Hacer una consulta al modelo
response = llm_chain.invoke("Cuantas horas de vuelo necesito para ser piloto privado en Argentina")
response
```
